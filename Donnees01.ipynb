```python
import csv

# Liste des fichiers à analyser
fichiers_a_lire = ['produits.csv', 'ventes.csv']

for nom_fichier in fichiers_a_lire:
    with open(nom_fichier, mode='r', encoding='utf-8') as fichier:
        lecteur = csv.reader(fichier)
        
        # Extraire la première ligne (les en-têtes / colonnes)
        colonnes = next(lecteur)
        
        # Compter le reste des lignes (les données)
        nombre_de_lignes = sum(1 for ligne in lecteur)
        
        # Affichage du résultat
        print(f"Fichier  : {nom_fichier}")
        print(f"Lignes   : {nombre_de_lignes}")
        print(f"Colonnes : {', '.join(colonnes)}")
        print("-" * 50)
```

```python
#La Boucle While repète tant que la condition est  verifiée..On l'emploie quand le nbre d'iterations n'est pas connu d'avance
stock = 10
commandes = [3,4,5,2]
servies = 0
while commandes and stock >= commandes[0]:
    stock -= commandes[0]
    commandes.pop(0)
    servies += 1
print(f"{servies} commandes servies, stock restant {stock}")
```

```python
#break et continue//Break quitte entièrement la boucle et Continue abandonne l'iteration en cours et passe directement à la suivante
montants = [ 120.0, 45.5, -3.0, 890.0, 61.2 ]
total = 0.0

for montant in montants:
    if montant < 0:
        continue             # ligne abberante,on l'ignore et continue
    if total > 1000:
        break               #objectif atteint,inutile de continuer
    total += montant
print(round(total, 2))

```

```python
#Une compréhension construit un conteneur en une expression,à partir d'un autre conteneur.Module 3 la suppose acquise
montants = [ 120.0, 45.5, 890.0, 61.2 ]

avec_tva = [ ]

for m in montants:
    avec_tva.append(round(m * 1.18, 2))
print(avec_tva)
```

```python
#version comprehension d'une liste en une ligne
avec_tva = [round(m * 1.18, 2) for m in montants]
print(avec_tva)
```

```python
def calculer_tva(montant_ht: float, taux_tva: float = 0.18) -> float:# saisir le montant et avoir automatique le prix TTC(Moi meme)
    return round(montant_ht * (1 + taux_tva), 2)

# Modifiez simplement cette valeur pour tester vos montants
montant = 1055550

montant_ttc = calculer_tva(montant)
print(f"Montant HT : {montant:.2f} | Montant TTC (18 %) : {montant_ttc:.2f}")
```
