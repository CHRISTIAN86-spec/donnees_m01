```python
import csv

# Liste des fichiers à analyser
fichiers_a_lire = ['produits.csv', 'ventes.csv']

for nom_fichier in fichiers_a_lire:
    with open(nom_fichier, mode='r', encoding='utf-8') as fichier:
        lecteur = csv.reader(fichier)
        
        # Extraire la première ligne (les en-têtes / colonnes)
        colonnes = next(lecteur)
        
        # Compter le reste des lignes (les données)
        nombre_de_lignes = sum(1 for ligne in lecteur)
        
        # Affichage du résultat
        print(f"Fichier  : {nom_fichier}")
        print(f"Lignes   : {nombre_de_lignes}")
        print(f"Colonnes : {', '.join(colonnes)}")
        print("-" * 50)
```
