```python
import csv

# Liste des fichiers à analyser
fichiers_a_lire = ['produits.csv', 'ventes.csv']

for nom_fichier in fichiers_a_lire:
    with open(nom_fichier, mode='r', encoding='utf-8') as fichier:
        lecteur = csv.reader(fichier)
        
        # Extraire la première ligne (les en-têtes / colonnes)
        colonnes = next(lecteur)
        
        # Compter le reste des lignes (les données)
        nombre_de_lignes = sum(1 for ligne in lecteur)
        
        # Affichage du résultat
        print(f"Fichier  : {nom_fichier}")
        print(f"Lignes   : {nombre_de_lignes}")
        print(f"Colonnes : {', '.join(colonnes)}")
        print("-" * 50)
```

```python
# Vérifier si un id_vente est répété
df_ventes.duplicated(subset=['id_vente']).sum()
```

```python
# Convertir la colonne pays en majuscules
df_ventes['pays'] = df_ventes['pays'].str.strip().str.upper()
df_ventes['pays'].unique()
```

```python
# 1. Remplacement des codes pays par leurs noms complets
df_ventes['pays'] = df_ventes['pays'].replace({
    'CI': "Cote dIvoire",
    'SN': "Senegal",
    'ML': "Mali",
    'CM': "Cameroun",
    'MA': "Maroc"

})

# 2. Conversion du type de la colonne mois en texte (string)
df_ventes['mois'] = df_ventes['mois'].astype(str)

# Vérification
print(df_ventes['pays'].unique())
print(df_ventes['mois'].dtype)
```

```python
df_ventes[['id_vente', 'pays', 'montant']]
```

```python
# Filtrer : Pays == 'Senegal' ET Montant > 500
ventes_cibles = df_ventes[(df_ventes['pays'] == 'Senegal') & (df_ventes['montant'] > 500)]
ventes_cibles
```

```python
# Trier par montant de façon décroissante (ascending=False)
df_trie = df_ventes.sort_values(by='montant', ascending=False)
df_trie.head()
```

```python
# Ventes effectuées soit en Cote dIvoire, soit au Maroc
df_ci_ma = df_ventes[(df_ventes['pays'] == 'Cote dIvoire') | (df_ventes['pays'] == 'Maroc')]
df_ci_ma.head()
```

```python
# Tri par pays (A-Z) puis par montant (décroissant)
df_multi_tri = df_ventes.sort_values(by=['pays', 'montant'], ascending=[True, False])
df_multi_tri.head(10)
```

```python
# 1. Filtre élégant avec .isin() pour la Côte d'Ivoire et le Maroc
df_ci_ma = df_ventes[df_ventes['pays'].isin(['Cote dIvoire', 'Maroc'])]

# 2. Tri par pays (A-Z) puis par montant (décroissant)
df_multi_tri = df_ventes.sort_values(by=['pays', 'montant'], ascending=[True, False])

# Afficher les 10 premières lignes du résultat trié
df_multi_tri.head(10)
```

```python
# Chiffre d'affaires total
df_ventes['montant'].sum()

```

```python
# Somme des montants par pays
ventes_par_pays = df_ventes.groupby('pays')['montant'].sum()
ventes_par_pays
```

```python
# Somme des montants et quantités par canal
ventes_par_canal = df_ventes.groupby('canal')[['montant', 'quantite']].sum()
ventes_par_canal
```

```python
# Calculer le chiffre d'affaires total, la moyenne et le nombre de ventes par pays
analyse_pays = df_ventes.groupby('pays')['montant'].agg(['sum', 'mean', 'count'])
analyse_pays
```

```python
# 1. Renommer les colonnes de l'agrégation
analyse_pays = analyse_pays.rename(columns={
    'sum': 'chiffre_affaires',
    'mean': 'vente_moyenne',
    'count': 'nb_transactions'
})

# 2. Trier par chiffre d'affaires décroissant
analyse_pays = analyse_pays.sort_values(by='chiffre_affaires', ascending=False)

analyse_pays
```

```python
# 1. Renommer les colonnes de l'agrégation
analyse_pays = analyse_pays.rename(columns={
    'sum': 'chiffre_affaires',
    'mean': 'vente_moyenne',
    'count': 'nb_transactions'
})

# 2. Trier par chiffre d'affaires décroissant
analyse_pays = analyse_pays.sort_values(by='chiffre_affaires', ascending=False)

analyse_pays
```

```python
# Groupement par pays ET par canal de vente
analyse_pays_canal = df_ventes.groupby(['pays', 'canal'])['montant'].agg(['sum', 'count'])

# Renommer les colonnes
analyse_pays_canal = analyse_pays_canal.rename(columns={
    'sum': 'chiffre_affaires',
    'count': 'nb_transactions'
})

analyse_pays_canal
```
